```python
import pandas as pd
import numpy as np
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.linear_model import LogisticRegression
from imblearn.over_sampling import SMOTE
import warnings
warnings.filterwarnings('ignore')

# 1. Load Data
df = pd.read_excel('kuesionardivisipramuka smaba.xlsx')

# 2. Preprocessing
kolom_dihapus = ['Nama Lengkap', 'Kelas']
for col in kolom_dihapus:
    if col in df.columns: df.drop(columns=[col], inplace=True)
target_col = 'Divisi yang paling di minati?'
df = df.fillna(df.mode().iloc[0])
fitur_kolom = [col for col in df.columns if col != target_col]

encoders = {}
df_encoded = df.copy()
for col in df.columns:
    if df[col].dtype == 'object':
        df_encoded[col] = df_encoded[col].astype(str).str.strip()  # <-- tambah ini
        le = LabelEncoder()
        df_encoded[col] = le.fit_transform(df_encoded[col])
        encoders[col] = le

X = df_encoded[fitur_kolom]
y = df_encoded[target_col]

# 3. Scaling & SMOTE
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
smote = SMOTE(random_state=42, k_neighbors=1)
X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

# 4. Model: Logistic Regression
model = LogisticRegression(max_iter=1000)
scores = cross_val_score(model, X_resampled, y_resampled, cv=5)

print(f"Model: Logistic Regression")
print(f"Akurasi Rata-rata: {scores.mean()*100:.2f}%")
```

```python
import pandas as pd
import numpy as np
import pickle
from sklearn.model_selection import cross_val_score
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.ensemble import RandomForestClassifier
from imblearn.over_sampling import SMOTE
import warnings
warnings.filterwarnings('ignore')

# ==========================================
# 1. LOAD DATA
# ==========================================
df = pd.read_excel('kuesionardivisipramuka smaba.xlsx')
print(f"Data Awal: {len(df)} baris")

# ==========================================
# 2. PREPROCESSING
# ==========================================
# Hapus kolom identitas
kolom_dihapus = ['Nama Lengkap', 'Kelas']
for col in kolom_dihapus:
    if col in df.columns:
        df.drop(columns=[col], inplace=True)

target_col = 'Divisi yang paling di minati?'

# Hapus baris yang targetnya kosong (jika ada)
df = df.dropna(subset=[target_col])

# Isi data kosong di fitur
df = df.fillna(df.mode().iloc[0])

fitur_kolom = [col for col in df.columns if col != target_col]

# Encoding (Teks -> Angka)
encoders = {}
df_encoded = df.copy()
for col in df.columns:
    if df[col].dtype == 'object':
        df_encoded[col] = df_encoded[col].astype(str).str.strip()  # <-- tambah ini
        le = LabelEncoder()
        df_encoded[col] = le.fit_transform(df_encoded[col])
        encoders[col] = le

X = df_encoded[fitur_kolom]
y = df_encoded[target_col]

# Scaling (Penting untuk SMOTE)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# SMOTE (Menambah Data Virtual agar Random Forest lebih pintar)
# random_state=42 agar hasilnya konsisten
smote = SMOTE(random_state=42, k_neighbors=1)
X_resampled, y_resampled = smote.fit_resample(X_scaled, y)

print(f"Data Setelah SMOTE: {len(y_resampled)} baris")

# ==========================================
# 3. TRAINING MODEL RANDOM FOREST
# ==========================================
# Parameter disesuaikan untuk data kecil:
# - n_estimators=100 (jumlah pohon)
# - max_depth=10 (dibatasi agar tidak overfitting)
# - min_samples_split=2 (minimum sampel untuk split)
model_rf = RandomForestClassifier(n_estimators=100, max_depth=10, random_state=42, min_samples_split=2)

# Evaluasi dengan Cross Validation
scores = cross_val_score(model_rf, X_resampled, y_resampled, cv=5, scoring='accuracy')

print("-" * 40)
print(f"Model: Random Forest")
print(f"Akurasi Rata-rata: {scores.mean()*100:.2f}%")
print("-" * 40)

# Latih model dengan seluruh data resampled
model_rf.fit(X_resampled, y_resampled)

# ==========================================
# 4. SIMPAN MODEL ARTIFACTS
# ==========================================
data_to_save = {
    'model': model_rf,
    'scaler': scaler,
    'encoders': encoders,
    'fitur_kolom': fitur_kolom,
    'target_col': target_col
}

filename = 'model_artifacts.pkl'
with open(filename, 'wb') as f:
    pickle.dump(data_to_save, f)

print(f"\n[SUCCESS] Model Random Forest berhasil disimpan ke '{filename}'")
print("File ini siap digunakan untuk aplikasi Streamlit Anda.")
```
